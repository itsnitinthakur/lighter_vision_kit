# Lighter inspection demo: the free replacement for LighterGrab, step by step

Every step of the pick-up station, shown as pictures:
1. a good lighter image  2. the learned outline and grab point  3. matching at any angle
4. front vs back side  5. defect detection  6. the numbers sent to the PLC  7. a whole folder run

The images here are **generated** (a body, a cap and a side button), because real lighter photos are not
collected yet. Replace them with photos from `D:\picture2` once you have them (last section).

Run: open this file in Jupyter or VS Code from the `replacement` folder, then *Run All*.

```python
import sys, math
from pathlib import Path
import cv2, numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, '.')
from lightergrab_cv import Recipe, learn, inspect, draw, transform, PlcLink, run_folder
from test_lightergrab_cv import lighter          # makes a synthetic lighter image

def show(*imgs, titles=None, size=4.5):
    fig, axes = plt.subplots(1, len(imgs), figsize=(size * len(imgs), size * 0.8))
    axes = np.atleast_1d(axes)
    for i, (ax, im) in enumerate(zip(axes, imgs)):
        ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im.ndim == 3 else im, cmap='gray')
        ax.set_title(titles[i] if titles else ''); ax.axis('off')
    plt.tight_layout(); plt.show()

print('OpenCV', cv2.__version__)
```

## 1. One good lighter, the learning image
LighterGrab learns from one good part. The operator draws the region and clicks the grab point; here the
grab point is 40 px left of the centre.

```python
good = lighter(320, 240, 0)
grab = (280, 240)
vis = good.copy()
cv2.drawMarker(vis, grab, (0, 200, 255), cv2.MARKER_CROSS, 30, 3)
show(good, vis, titles=['good lighter (learning image)', 'grab point chosen by the operator'])
```

## 2. What the software learns: the outline
Edges are found (Canny), the outer outline is kept, and its points are stored relative to the outline's
centre at angle 0. This is the free equivalent of LighterGrab's `contourN.data`.

```python
recipe = learn(good, grab, Recipe(name='demo'))
gray = cv2.cvtColor(good, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 0), 30, 90)
outline = good.copy()
pts = np.array(recipe.contour)
m = cv2.moments(cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)[0][0])
cx, cy = m['m10'] / m['m00'], m['m01'] / m['m00']
for x, y in transform(pts, cx, cy, 0).astype(int):
    cv2.circle(outline, (x, y), 2, (0, 255, 0), -1)
show(edges, outline, titles=['edges (Canny)', f'learned outline: {len(pts)} points'])
print('grab point relative to centre:', recipe.grab_point, '| area:', round(recipe.area), 'px')
```

## 3. Matching at any angle
The same lighter turned to six angles. Each result: green cross = grab point moved to the lighter's real
position, arrow = angle, OK/NG + side.

```python
angles = [0, 37, 90, 145, -120, 179]
imgs, titles = [], []
for a in angles:
    img = lighter(330, 250, a)
    res = inspect(img, recipe)
    imgs.append(draw(img, res))
    r = res[0]
    titles.append(f'true {a} deg -> found {r.angle:.1f} deg, score {r.score:.2f}')
show(*imgs[:3], titles=titles[:3]); show(*imgs[3:], titles=titles[3:])
```

## 4. Front side vs back side
A lighter lying the other way up shows the **mirrored** outline (the button is on the other side). The
software fits both and reports `front` or `back`, which the original sends to the PLC so the robot can flip it.

```python
front = lighter(330, 250, 30)
back = lighter(330, 250, 30, back=True)
rf, rb = inspect(front, recipe)[0], inspect(back, recipe)[0]
show(draw(front, [rf]), draw(back, [rb]), titles=[f'side = {rf.side}', f'side = {rb.side}'])
```

## 5. Defect detection (the "max defect chain length" rule)
The learned outline is laid over the found lighter. Points farther than 3 px from the real edge are
**bad** (red). If more than 16 bad points sit in a row (LighterGrab's `maxDefectChainLength`), it is NG.

```python
def defect_view(img, r):
    g = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    e = cv2.Canny(cv2.GaussianBlur(g, (5, 5), 0), 30, 90)
    e = cv2.morphologyEx(e, cv2.MORPH_CLOSE, np.ones((5, 5), np.uint8))
    c = max(cv2.findContours(e, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)[0], key=cv2.contourArea)
    m = cv2.moments(c); cx, cy = m['m10'] / m['m00'], m['m01'] / m['m00']
    t = np.array(recipe.contour) * ([1, -1] if r.side == 'back' else [1, 1])
    placed = transform(t, cx, cy, r.angle)
    canvas = np.full(g.shape, 255, np.uint8); cv2.drawContours(canvas, [c], -1, 0, 1)
    dist = cv2.distanceTransform(canvas, cv2.DIST_L2, 3)
    out = img.copy()
    for x, y in placed.astype(int):
        bad = dist[min(max(y, 0), g.shape[0] - 1), min(max(x, 0), g.shape[1] - 1)] > recipe.defect_tolerance_px
        cv2.circle(out, (x, y), 2, (0, 0, 255) if bad else (0, 255, 0), -1)
    return out

ok_img, ng_img = lighter(320, 240, 20), lighter(320, 240, 20, broken=True)
r_ok, r_ng = inspect(ok_img, recipe)[0], inspect(ng_img, recipe)[0]
show(defect_view(ok_img, r_ok), defect_view(ng_img, r_ng),
     titles=[f'OK: longest bad run {r_ok.defect_chain} px', f'NG: longest bad run {r_ng.defect_chain} px ({r_ng.reason})'])
```

## 6. What goes to the PLC
Dry run: nothing is sent. These are the six 16-bit values the code would write. **The register layout is a
placeholder** until the real map comes from the PLC program (`.pdw`).

```python
plc = PlcLink(None)          # None = dry run, no network
rows = []
for name, r in [('front OK', rf), ('back OK', rb), ('broken', r_ng)]:
    regs = plc.send(r)
    rows.append((name, r.ok, round(r.x, 1), round(r.y, 1), round(r.angle, 1), r.side, regs))
print(f"{'case':10} {'ok':5} {'x':>7} {'y':>7} {'angle':>7} {'side':6} registers [flag, X*10, Y*10, angle*10, side, counter]")
for row in rows:
    print(f'{row[0]:10} {str(row[1]):5} {row[2]:7} {row[3]:7} {row[4]:7} {row[5]:6} {row[6]}')
```

## 7. A whole folder, the way it runs on the line
Writes `results.csv` (one row per lighter, which LighterGrab never saved) and an image of every NG.

```python
import pandas as pd, shutil
demo = Path('demo_images'); shutil.rmtree(demo, ignore_errors=True); demo.mkdir()
cases = [(30, 0, 0), (120, 0, 0), (-60, 1, 0), (10, 1, 0), (75, 0, 1), (160, 0, 1)]
for i, (a, b, br) in enumerate(cases):
    cv2.imwrite(str(demo / f'lighter_{i}.png'), lighter(330, 250, a, back=bool(b), broken=bool(br)))
out = Path('demo_output'); shutil.rmtree(out, ignore_errors=True)
counts = run_folder(recipe, demo, out, PlcLink(None))
print(counts)
pd.read_csv(out / 'results.csv')
```

```python
ng = sorted((out / 'ng').glob('*.png'))
show(*[cv2.imread(str(p)) for p in ng], titles=[p.name for p in ng])
```

## 8. Your real images (after the factory visit)
Put one good lighter photo and a folder of photos next to this notebook, set the grab point in pixels,
and run this cell. Flip settings come from the converted recipe (`tools/toolparam_to_recipe.py`).

```python
REAL_GOOD = Path('real/good.png')          # one good lighter from D:\\picture2
REAL_FOLDER = Path('real/batch')           # a folder of photos to check
REAL_GRAB = (280, 240)                     # grab point in pixels on REAL_GOOD

if REAL_GOOD.exists():
    real_recipe = learn(cv2.imread(str(REAL_GOOD)), REAL_GRAB, Recipe(name='real'))
    for p in sorted(REAL_FOLDER.glob('*'))[:6]:
        img = cv2.imread(str(p))
        if img is not None:
            show(draw(img, inspect(img, real_recipe)), titles=[p.name])
else:
    print('No real images yet: copy them into real/ after the visit.')
```
